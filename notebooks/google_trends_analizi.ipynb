## 1- Kütüphanelerin İçe Aktarılması

 Bu aşamada veri analizi ve görselleştirme işlemlerinde kullanacağımız Python kütüphanelerini içe aktaracağız.

- Pandas: Verilerin okunması, düzenlenmesi ve analiz edilmesi.

Bu kütüphaneler, veri setinin yapısını incelememize ve sonraki analiz adımlarını gerçekleştirmemize yardımcı olacaktır.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Grafiklerin görünümünü düzenleyelim.
sns.set_theme(style="whitegrid")

# Japonca ve diğer Unicode kategori adlarının desteklenen yazı tipleriyle çizilmesini dene.
plt.rcParams["font.family"] = ["Noto Sans CJK JP", "DejaVu Sans", "sans-serif"]
```

## 2-Veri Setinin İçeri Aktarılması

```python

df = pd.read_csv("trends.csv")

# Veri setinin ilk beş satırını görüntüleyelim.
display(df.head())
```

## 3-Veri setinin boyutunu ve sütunlarını inceleme

```python

# Veri setindeki satır ve sütun sayısını görüntüleme
print("Satır ve sütun sayısı:", df.shape)

# Veri setindeki sütun isimlerini listeleme
print("\nSütun isimleri:")
print(df.columns.tolist())

# Sütunların veri türlerini ve dolu değer sayılarını inceleme
print("\nVeri setinin genel bilgileri:")
df.info()
```

## 4-Eksik Verilerin Kontrolü

```python

eksik_degerler = df.isnull().sum()

print("Sütunlara göre eksik değer sayıları:")
display(eksik_degerler.to_frame(name="Eksik Değer Sayısı"))

print("Toplam eksik değer sayısı:", df.isnull().sum().sum())
```

 ## 5-Yinelenen Kayıtların Kontrol Edilmesi

```python

yinelenen_sayi = df.duplicated().sum()

print("Yinelenen satır sayısı:", yinelenen_sayi)
```

## 6-Yinelenen Kayıtları İnceleme

```python

yinelenen_kayitlar = df[df.duplicated(keep=False)]

display(yinelenen_kayitlar.sort_values(
    by=["location", "year", "category", "query"]
))
```

## 7-Yinelenen Kayıtların Temizlenmesi

```python

# Temizleme öncesindeki satır sayısını kaydetme
onceki_satir_sayisi = len(df)

# Tamamen aynı olan yinelenen satırları silme
df = df.drop_duplicates().reset_index(drop=True)

# Temizleme sonrasındaki satır sayısını hesaplama
sonraki_satir_sayisi = len(df)

# Temizleme işleminin sonuçlarını görüntüleme
print("Temizleme öncesi satır sayısı:", onceki_satir_sayisi)
print("Kaldırılan yinelenen satır sayısı:", onceki_satir_sayisi - sonraki_satir_sayisi)
print("Temizleme sonrası satır sayısı:", sonraki_satir_sayisi)
print("Kalan yinelenen satır sayısı:", df.duplicated().sum())
```

## 8-Kategorik Değişkenlerin incelenmesi

```python

# Kategorik değişkenlerdeki farklı değer sayılarını hesaplama
print("Farklı location sayısı:", df["location"].nunique())
print("Farklı kategori sayısı:", df["category"].nunique())
print("Farklı arama terimi sayısı:", df["query"].nunique())

# En sık karşılaşılan coğrafi kapsamları listeleme
print("\nEn sık görülen location değerleri:")
display(df["location"].value_counts().head(10).to_frame("Kayıt Sayısı"))

# En sık karşılaşılan kategorileri listeleme
print("\nEn sık görülen kategoriler:")
display(df["category"].value_counts().head(10).to_frame("Kayıt Sayısı"))
```

## 9-Yılların ve Coğrafi Kapsamların İncelenmesi


```python

# Veri setindeki en erken ve en geç yılları bulma
print("Başlangıç yılı:", df["year"].min())
print("Bitiş yılı:", df["year"].max())

# Her yıldaki kayıt sayısını inceleme
print("\nYıllara göre kayıt sayıları:")
display(
    df["year"]
    .value_counts()
    .sort_index()
    .to_frame(name="Kayıt Sayısı")
)

# Coğrafi kapsamları alfabetik olarak listeleme
print("\nCoğrafi kapsam örnekleri:")
display(
    pd.DataFrame({
        "Location": sorted(df["location"].unique())
    }).head(20)
)

# Küresel kayıtların sayısını kontrol edelim.
print(
    "\nGlobal kayıt sayısı:",
    (df["location"] == "Global").sum()
)
```

## 10-Yıllara Göre Kayıt Dağılımının Görselleştirilmesi

```python

# Her yıl için kayıt sayısını hesaplama
yillik_kayit = df["year"].value_counts().sort_index()

# Yıllara göre kayıt sayılarını çubuk grafikle gösterme
plt.figure(figsize=(14, 6))
yillik_kayit.plot(kind="bar", color="mediumpurple", edgecolor="black")

plt.title("Veri Setindeki Kayıt Sayısı Yıllara Göre Değişiyor", fontsize=15)
plt.xlabel("Yıl")
plt.ylabel("Kayıt Sayısı")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 11-En Sık Görülen Arama Kategorilerinin Analizi

```python

# Her kategoride bulunan kayıt sayısını hesaplama
kategori_sayilari = df["category"].value_counts().head(10)

# En sık görülen 10 kategoriyi yatay çubuk grafikle gösterelim.
plt.figure(figsize=(12, 7))

kategori_sayilari.sort_values().plot(
    kind="barh",
    color="cornflowerblue",
    edgecolor="black"
)

plt.title("Veri Setinde En Çok Temsil Edilen 10 Kategori", fontsize=15)
plt.xlabel("Kayıt Sayısı")
plt.ylabel("Kategori")
plt.tight_layout()
plt.show()
```

## 12-Yıllara Göre Kategori Değişimi


Bu aşamada People, Movies, TV Shows ve Songs kategorilerinin yıllara göre kayıt sayıları karşılaştırılacaktır. Böylece bu kategorilerin veri setindeki temsilinin zaman içinde nasıl değiştiği incelenecektir.

Karşılaştırma, gerçek arama hacmini değil, veri setinde bulunan kayıtların dağılımını gösterecektir.


```python

# Karşılaştırmak istediğimiz kategorileri belirleyelim.
secili_kategoriler = ["People", "Movies", "TV Shows", "Songs"]

# Seçilen kategorilerin yıllara göre kayıt sayılarını hesaplayalım.
kategori_yil = (
    df[df["category"].isin(secili_kategoriler)]
    .groupby(["year", "category"])
    .size()
    .unstack(fill_value=0)
)

# Kategorilerin yıllara göre değişimini çizgi grafikle gösterelim.
kategori_yil.plot(figsize=(14, 7), marker="o")

plt.title("Seçilen Kategorilerin Yıllara Göre Kayıt Dağılımı", fontsize=15)
plt.xlabel("Yıl")
plt.ylabel("Kayıt Sayısı")
plt.legend(title="Kategori")
plt.xticks(sorted(df["year"].unique()), rotation=45)
plt.tight_layout()
plt.show()
```

## 13-Küresel Arama Trendlerinin İncelenmesi


```python

# Yalnızca Global kapsamındaki kayıtları seçme
global_veri = df[df["location"] == "Global"].copy()

# Global kayıtların sayısını kontrol etme
print("Global kayıt sayısı:", len(global_veri))

# Global veride en sık görülen kategorileri listeleme
print("\nGlobal veride en sık görülen kategoriler:")
display(
    global_veri["category"]
    .value_counts()
    .head(10)
    .to_frame(name="Kayıt Sayısı")
)

# Global verideki arama terimlerini sıralamalarına göre inceleme
print("\nGlobal veride ilk sıralarda yer alan arama terimleri:")
display(
    global_veri.sort_values(["year", "rank"])
    [["year", "category", "rank", "query"]]
    .head(20)
)
```

## 14-Küresel Trend Kayıtlarının Yıllara Göre Dağılımı


```python

# Global kayıtları yıllara göre sayma
global_yillik = global_veri["year"].value_counts().sort_index()

# Yıllara göre kayıt dağılımını çizgi grafikle gösterme
plt.figure(figsize=(13, 6))

plt.plot(
    global_yillik.index,
    global_yillik.values,
    marker="o",
    linewidth=2,
    color="darkorange"
)

plt.title("Global Trend Kayıtlarının Yıllara Göre Dağılımı", fontsize=15)
plt.xlabel("Yıl")
plt.ylabel("Kayıt Sayısı")
plt.xticks(sorted(global_veri["year"].unique()), rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 15-Global Verilerde Birinci Sıradaki Arama Terimleri


```python

# Global veride birinci sırada yer alan kayıtları seçme
global_birinciler = global_veri[global_veri["rank"] == 1].copy()

# Kayıtları yıl ve kategoriye göre sıralayalım.
global_birinciler = global_birinciler.sort_values(
    ["year", "category"]
)

# İlk sıradaki arama terimlerini görüntüleme
display(
    global_birinciler[["year", "category", "query"]]
    .reset_index(drop=True)
)

# Kaç farklı yıl-kategori birincisi bulunduğunu gösterme
print("Birinci sıradaki kayıt sayısı:", len(global_birinciler))
```

## 16-Birden Fazla Yıl Birinci Sırada Yer Alan Arama Terimleri


```python

# Birinci sıradaki arama terimlerinin kaç kez yer aldığını hesaplayalım.
terim_tekrar = global_birinciler["query"].value_counts()

# Birden fazla kez görülen ilk 15 terimi seçelim.
tekrar_eden_terimler = terim_tekrar[terim_tekrar > 1].head(15)

# Sonuçları görüntüleyelim.
print("Birden fazla kez birinci sırada görülen terimler:")
display(tekrar_eden_terimler.to_frame(name="Birinci Sıra Kayıt Sayısı"))

# Sonuçları yatay çubuk grafikle gösterelim.
if not tekrar_eden_terimler.empty:
    tekrar_eden_terimler.sort_values().plot(
        kind="barh",
        figsize=(10, 6),
        color="teal"
    )

    plt.title("Birden Fazla Kez Birinci Sırada Yer Alan Arama Terimleri")
    plt.xlabel("Kayıt Sayısı")
    plt.ylabel("Arama Terimi")
    plt.tight_layout()
    plt.show()
else:
    print("Birden fazla kez birinci sırada görülen terim bulunamadı.")
```

## 17-Tekrar Eden Arama Terimlerinin Yıl ve Kategori Bazında İncelenmesi


```python

# Önceki analizde öne çıkan terimleri belirleyelim.
incelenecek_terimler = [
    "Britney Spears",
    "The Simpsons",
    "American Idol",
    "Eminem",
    "Whitney Houston",
    "World Cup"
]

# Seçilen terimlerin yıl, kategori ve sıralama bilgilerini görüntüleyelim.
terim_detaylari = global_birinciler[
    global_birinciler["query"].isin(incelenecek_terimler)
].sort_values(["query", "year", "category"])

display(
    terim_detaylari[["query", "year", "category", "rank"]]
    .reset_index(drop=True)
)
```

## 18-Öne Çıkan Terimlerin Yıllara Göre Dağılımı


```python
terim_yil = (
    global_birinciler[
        global_birinciler["query"].isin(incelenecek_terimler)
    ]
    .groupby(["year", "query"])
    .size()
    .unstack(fill_value=0)
)

terim_yil.plot(
    kind="line",
    figsize=(12, 6),
    marker="o"
)

plt.title("Öne Çıkan Terimlerin Yıllara Göre Birincilik Kayıtları")
plt.xlabel("Yıl")
plt.ylabel("Birincilik kayıt sayısı")
plt.xticks(range(2001, 2021, 2))
plt.legend(title="Arama terimi", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

## 19-Ülkelerin Trend Kategorilerinin Karşılaştırılması


```python

# Global kaydını çıkarıp en fazla kaydı bulunan 5 ülkeyi seçiyoruz.
ulke_verisi = df[df["location"] != "Global"].copy()

ilk_bes_ulke = (
    ulke_verisi["location"]
    .value_counts()
    .head(5)
    .index
)

ulke_kategori = (
    ulke_verisi[ulke_verisi["location"].isin(ilk_bes_ulke)]
    .groupby(["location", "category"])
    .size()
    .reset_index(name="kayit_sayisi")
)

# Her ülkenin en sık görülen 5 kategorisini alıyoruz.
ilk_kategoriler = (
    ulke_kategori
    .sort_values(["location", "kayit_sayisi"], ascending=[True, False])
    .groupby("location")
    .head(5)
)

display(ilk_kategoriler.sort_values(["location", "kayit_sayisi"],
                                    ascending=[True, False]))
```

## 20-Ülkelere Göre Öne Çıkan Trend Kategorileri


```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(13, 7))

sns.barplot(
    data=ilk_kategoriler,
    x="kayit_sayisi",
    y="category",
    hue="location"
)

plt.title("Ülkelere Göre Öne Çıkan Trend Kategorileri")
plt.xlabel("Veri setindeki kayıt sayısı")
plt.ylabel("Kategori")
plt.legend(title="Ülke", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python

grafik_verisi = ilk_kategoriler.copy()

grafik_verisi["etiket"] = (
    grafik_verisi["location"] + " — " + grafik_verisi["category"]
)

grafik_verisi = grafik_verisi.sort_values("kayit_sayisi")

plt.figure(figsize=(12, 10))

plt.barh(
    grafik_verisi["etiket"],
    grafik_verisi["kayit_sayisi"]
)

plt.title("Ülke ve Kategori Birleşimlerine Göre Kayıt Sayıları")
plt.xlabel("Veri setindeki kayıt sayısı")
plt.ylabel("Ülke — kategori")
plt.tight_layout()
plt.show()
```
